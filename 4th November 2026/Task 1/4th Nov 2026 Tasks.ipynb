B. Filter, clean and transform

```python
from pyspark.sql.functions import col
bookings_df.filter(col("booking_status")=="Confirmed").show()
```

```python
bookings_df.filter((col("tickets")>3)&(col("price_per_ticket")>200)).show()
```

```python
from pyspark.sql.functions import sum,when
bookings_df.select(
    sum(when(col("discount_pct").isNull(),1).otherwise(0)).alias("missing discount"),
    sum(when(col("payment_method").isNull(),1).otherwise(0)).alias("missing payment method")
).show()
```

```python
cleaned_df=bookings_df.fillna({
    "discount_pct":0,
    "payment_method":"Unknown"
})
display(cleaned_df)
```

```python
cleaned_df=cleaned_df.withColumn("gross_amount",col("price_per_ticket")*col("tickets"))
display(cleaned_df)
```

```python
cleaned_df=cleaned_df.withColumn(
    "net_amount",
    col("gross_amount")*(1-col("discount_pct")/100)
)
display(cleaned_df)
```

```python
cleaned_df=cleaned_df.withColumn(
    "book_size",
    when(col("tickets")>=4,"Large").otherwise("Regular")
)
display(cleaned_df)
```

```python
cleaned_df.orderBy(col("net_amount").desc()).show(5)
```

C. Aggregations and joins

```python
cleaned_df.groupBy("booking_status").count().show()
```

```python
cleaned_df.filter(col("booking_status")=="Confirmed").groupBy("movie_title").agg(
    sum("tickets").alias("total_tickets"),
    sum("net_amount").alias("total_revenue")
).show()
```

```python
cleaned_df.filter(col("booking_status")=="Confirmed").groupBy("cinema_id").agg(
    sum("net_amount").alias("total_revenue")
).orderBy(col("total_revenue").desc()).show(3)
```

```python
joined_df.select("cinema_name", "location.city", "movie_title" ,"booking_status").show()
```

```python
joined_df.filter(col("booking_status")=="Confirmed").groupBy(col("location.city")).agg(
    sum("net_amount").alias("net_revenue")
).show()
joined_df.filter(col("cinema_name").isNull()).show()
```

D. JSON, dates and DataFrame quality

```python
cinemas_df.select(col("cinema_name"),col("location.zone")).show()
```

```python
from pyspark.sql.functions import explode
cinemas_df.select("cinema_name",explode("facilities").alias("facility")).show()
```

```python
from pyspark.sql.functions import to_date,date_diff
date_df=cleaned_df.withColumn("booking_date",to_date(col("booking_date")))\
    .withColumn("show_date",to_date(col("show_date")))
date_df=date_df.withColumn("days_to_show",date_diff(col("show_date"),col("booking_date")))
display(date_df)
```

```python
bookings_df.groupBy("booking_id").count().filter(col("count")>1).show()
nodup_df=cleaned_df.dropDuplicates(["booking_id"])
display(nodup_df)
```

E. Window functions and Delta

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank
movie_revenue_df=cleaned_df.filter(col("booking_status")=="Confirmed").groupBy("movie_title").agg(
    sum("net_amount").alias("total_revenue")
)
window=Window.orderBy(col("total_revenue").desc())
ranked_df=movie_revenue_df.withColumn("movie_rank",rank().over(window))
display(ranked_df)
```

```python
final_df = nodup_df.join(
    cinemas_df,
    on="cinema_id",
    how="left"
)

final_df.write.format("delta").mode("overwrite").saveAsTable(
    "cinema_booking_analytics"
)
```

Final challenge

```python
final_cleaned_df = bookings_df.fillna({
    "discount_pct": 0,
    "payment_method": "Unknown"
}).dropDuplicates(["booking_id"])
final_cleaned_df = final_cleaned_df.withColumn(
    "gross_amount",
    col("tickets") * col("price_per_ticket")
).withColumn(
    "net_amount",
    col("gross_amount") * (1 - col("discount_pct") / 100)
)
final_report_df = final_cleaned_df.filter(
    col("booking_status") == "Confirmed"
).join(
    cinemas_df,
    on="cinema_id",
    how="left"
)
report_df = final_report_df.groupBy(
    col("location.city"),
    col("movie_title")
).agg(
    sum("tickets").alias("total_tickets"),
    sum("net_amount").alias("total_revenue")
)
display(report_df)
```

```python
report_df.write.format("delta").mode("overwrite").saveAsTable( "cinema_performance_report");
```
