Part A — Retrieving and filtering data

```python
%sql select * from pattu_databricks.default.fitness_memberships_demo
```

```python
%sql
select member_name,city,membership_plan from fitness_memberships_demo;
```

```python
%sql
select * from fitness_memberships_demo where city="Hyderabad";
```

```python
%sql
select * from fitness_memberships_demo where monthly_fee>2000;
```

```python
%sql
select distinct city from fitness_memberships_demo;
```

```python
%sql
select * from fitness_memberships_demo where membership_plan in ("Premium","Standard");
```

```python
%sql
select * from fitness_memberships_demo where monthly_fee between 1000 and 2000;
```

```python
%sql
select * from fitness_memberships_demo where member_name like "S%";
```

```python
%sql
select * from fitness_memberships_demo where city="Hyderabad" and status="Active";
```

```python
%sql
select member_name as Name ,monthly_fee as  Fee from fitness_memberships_demo;
```

```python
%sql
select * from fitness_memberships_demo as f where f.membership_plan="Premium";
```
