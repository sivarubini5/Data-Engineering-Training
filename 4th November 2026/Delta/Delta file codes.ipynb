Parquet file--> column oriented ,used for analytics

 Delta file -->built on top of parquet file as it doesnt have acid props,transaction history and some other things.

```python
%sql
CREATE OR REPLACE TABLE office_inventory_demo (
    product_id INT,
    product_name STRING,
    category STRING,
    city STRING,
    price INT,
    quantity INT
)
USING DELTA;
```
