Section E — MERGE, UPSERT and SCD

```python
%sql
SELECT i.registration_id,
       CASE
           WHEN w.registration_id IS NULL THEN 'New'
           ELSE 'Existing'
       END AS registration_type
FROM incoming_registrations i
LEFT JOIN event_registrations_work w
ON i.registration_id = w.registration_id;
```

```python
%sql
MERGE INTO event_registrations_work AS target
USING incoming_registrations AS source
ON target.registration_id = source.registration_id

WHEN MATCHED THEN
UPDATE SET
    target.event_id = source.event_id,
    target.attendee_name = source.attendee_name,
    target.registration_date = source.registration_date,
    target.ticket_type = source.ticket_type,
    target.tickets = source.tickets,
    target.unit_price = source.unit_price,
    target.status = source.status,
    target.channel = source.channel

WHEN NOT MATCHED THEN
INSERT (
    registration_id,
    event_id,
    attendee_name,
    registration_date,
    ticket_type,
    tickets,
    unit_price,
    status,
    channel
)
VALUES (
    source.registration_id,
    source.event_id,
    source.attendee_name,
    source.registration_date,
    source.ticket_type,
    source.tickets,
    source.unit_price,
    source.status,
    source.channel
);
```

```python
%sql
CREATE OR REPLACE VIEW event_revenue_view AS
SELECT e.event_name,
       e.city,
       SUM(r.tickets * r.unit_price) AS total_confirmed_revenue
FROM event_registrations_source r
JOIN event_master e
ON r.event_id = e.event_id
WHERE r.status = 'Confirmed'
  AND r.tickets > 0
  AND r.unit_price > 0
GROUP BY e.event_name, e.city;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW pending_registrations AS
SELECT *
FROM event_registrations_source
WHERE status = 'Pending';
```

Section I — Structured Streaming

```python
rate_df = spark.readStream \
    .format("rate") \
    .option("rowsPerSecond", 5) \
    .load()
```

```python
rate_df.writeStream \
    .format("console") \
    .outputMode("append") \
    .option("checkpointLocation", "/Volumes/workspace/default/streaming_checkpoints/rate_console_demo") \
    .trigger(availableNow=True) \
    .start()
```

```python
from pyspark.sql.functions import col

transaction_df = rate_df.withColumn(
    "transaction_id",
    col("value")
).withColumn(
    "transaction_amount",
    col("value") * 100
)
```

```python
filtered_df = transaction_df.filter(
    col("transaction_amount") > 500
)
```

```python
from pyspark.sql.functions import when, col

categorized_df = filtered_df.withColumn(
    "transaction_category",
    when(col("transaction_amount") >= 1000, "High")
    .otherwise("Regular")
)

aggregated_df = categorized_df.groupBy(
    "transaction_category"
).count()

query = aggregated_df.writeStream \
    .format("console") \
    .outputMode("complete") \
    .option("checkpointLocation","/Volumes/workspace/default/streaming_checkpoints/events_parquet") \
    .trigger(availableNow=True) \
    .start()

query.awaitTermination()
```
